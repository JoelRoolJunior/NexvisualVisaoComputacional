# Imports

```python
import numpy as np
import easyocr
import matplotlib.pyplot as plt
from PIL import Image
```

```python
eng_img = Image.open(r"images\eng.jpeg")
num_img = Image.open(r"images\num.jpeg")


fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(16,9))
axes[0].imshow(eng_img)
axes[1].imshow(num_img)
```

# Carregar modelo e utilizar modelo

```python
reader = easyocr.Reader(['en']) # carrega o modelo de inglês para a memoria
```

```python
plt.imshow(eng_img)
result = reader.readtext(r'images\eng.jpeg', detail=0)
print(result)
```

```python
plt.imshow(num_img)
result = reader.readtext(r'images\num.jpeg', detail=0)
print(result)
```
