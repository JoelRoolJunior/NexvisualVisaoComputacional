## Esquemas de interpolação de matrizes

```python
img = Image.open(r'images\os-insetos-caracterizam-se-pela-presenca-tres-pares-pernas-um-par-antenas-576807a97b345.webp')
img.thumbnail((64,64)) # redimensionamento da imagem
imgplot = plt.imshow(img)
```

```python
# interpolação bilinear
imgplot = plt.imshow(img, interpolation='bilinear')
```

```python
# interpolação bicúbica
imgplot = plt.imshow(img, interpolation='bicubic')
```
