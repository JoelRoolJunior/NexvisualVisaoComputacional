```python
from ultralytics import YOLO
import cv2
import os
import torch
```

### Processar dados

```python
# Baixa o modelo
model = YOLO('yolo11n.pt')# Baixe o modelo padrão da YOLO, sugestão yolo11n

# Jogao o modelo para o dispositivo de computação
device = 'cuda' if torch.cuda.is_available() else 'cpu' # defina o dispositivo de computação
model = model.to(device) # Mova o modelo para o dispositivo

# processa o vídeo
results = model.track(source="CarsTrafic.mp4", conf=0.1, iou=0.7)
```

### Salvar Vídeo

```python
# validar resultados
for r in results:
    print(r)
```

```python
# Cria o escritor do vídeo de output
# Você pode pesquisar para salvar em outros formatos
results = list(results)

first_frame = results[0].plot()
h, w, _ = first_frame.shape

fourcc = cv2.VideoWriter_fourcc(*'mp4v')

writer = cv2.VideoWriter('output.mp4', fourcc, 20, (w, h))

# Escreve o vídeo de output
for r in results:
    frame = r.plot()
    writer.write(frame)

writer.release()
```
