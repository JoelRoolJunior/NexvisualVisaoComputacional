# Imports

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2 as cv
import torch
import torchvision
import ultralytics
```

```python
image = cv.imread(r'images\street.jpeg')

plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

# Segmentação Semãntica

```python
# carregar o modelo SAM
model = ultralytics.SAM('sam_b.pt')

# Roda a inferência do modelo SAM.
results = model(r'images\street.jpeg')

for i, result in enumerate(results):
    result.save(filename=f'results_{i}.jpg', boxes=False) # Salva os resultados em uma imagem no disco

# carrega a imagem que acabou de ser salva para visualizarmos
results_image = cv.imread('results_0.jpg')

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(19,9))
axes[0].imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
axes[1].imshow(cv.cvtColor(results_image, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```
