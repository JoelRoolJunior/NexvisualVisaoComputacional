# Imports

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2 as cv
import torch
import torchvision
import ultralytics
```

```python
image = cv.imread(r'images\street.jpeg')
```

```python
plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

# Detecção de Objetos

```python
# carregar o modelo pre treinado YOLO11n
model = ultralytics.YOLO('yolo11n.pt')

# Roda a inferência do modelo YOLO
results = model(image)
for i, result in enumerate(results):
    result.save(filename=f'results_{i}.jpg') # salva os resultados em uma imagem no disco

results_image = cv.imread('results_0.jpg')
plt.imshow(cv.cvtColor(results_image, cv.COLOR_BGR2RGB))
```
