# Importações

```python
import numpy as np
```

# Operações Básicas com Vetores

```python
# Criando vetores
u = np.array([1,3])
v = np.array([3,1])
```

```python
# Soma  e Subtração
y = u + v
w = u - v

# Divisão e Multiplicação
a = u / v
b = u * v

# Produto escalar
alpha = 2
escalar = u * alpha

# Produto interno
interno = np.dot(u,v)

# Produto vetorial
u3 = np.array([1,3,1])
v3 = np.array([3,1,1])
vetorial = np.cross(u3,v3)
```

```python
print(f'Soma: {y}')
print(f'Subtração: {w}')
print(f'Divisão: {a}')
print(f'Multiplicação: {b}')
print(f'Produto Escalar: {escalar}')
print(f'Produto Interno: {interno}')
print(f'Produto Vetorial: {vetorial}')
```
