# Importações

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2 as cv
```

# Ler imagem do Disco

```python
image = cv.imread(r'images\os-insetos-caracterizam-se-pela-presenca-tres-pares-pernas-um-par-antenas-576807a97b345.webp')
```

As cores são trocadas no OpenCV, ela utiliza o padrão BGR ao invés de RGB

```python
plt.imshow(image)
```

# Transformações de Imagens

```python
gain = np.clip(image / 255 * 1.001, 0 , 1)
plt.imshow(gain)
```

Podemos converter os espaço de cores para RGB utilizando o OpenCV

```python
img_rgb = cv.cvtColor(image, cv.COLOR_BGR2RGB)
plt.imshow(img_rgb)
```

podemos recortar a imagem

```python
crop = img_rgb[0:250, 0:250]
plt.imshow(crop)
```

Podemos Redimensionar imagens

```python
larger = cv.resize(img_rgb, (int(image.shape[1] * 4), int(image.shape[0] * 4)))
plt.imshow(larger)
```
