# Importações

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2 as cv
```

# Leitura da Imagem

```python
image = cv.imread(r'images\vertical-shot-european-bee-eater-tree-branch-sunlight.jpg')
image_gray = cv.cvtColor(image, cv.COLOR_BGR2GRAY)
plt.imshow(image_gray,cmap='gray')
```

# Convoluções

```python
x_canny = np.array([[-1,0,1], [-2,0,2], [-1,0,1]])
y_canny = np.array([[-1,-2,-1], [0,0,0], [1,2,1]])

sharp = np.array([[0,-1,0], [-1,5,-1], [0,-1,0]])

print(f'x carry kernel:\n {x_canny} \n')
print(f'y carry kernel:\n {y_canny} \n')
print(f'sharp carry kernel:\n {sharp} \n')
```

```python
image_canny = cv.filter2D(image_gray, -1,x_canny)
image_canny = cv.filter2D(image_gray, -1,y_canny)
plt.imshow(image_canny, cmap='gray')
```

```python
image_sharp = cv.filter2D(image,-1, sharp)
fig,axes = plt.subplots(nrows=1, ncols=2, figsize=(10,8))
axes[0].imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))
axes[1].imshow(cv.cvtColor(image_sharp, cv.COLOR_BGR2RGB))
```
