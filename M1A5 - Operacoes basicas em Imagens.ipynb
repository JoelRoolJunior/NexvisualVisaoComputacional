# Importações

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2 as cv
```

# Transformação de Imagens

```python
image = cv.imread(r'images\vertical-shot-european-bee-eater-tree-branch-sunlight.jpg')
image_rgb = cv.cvtColor(image, cv.COLOR_BGR2RGB)
plt.imshow(image_rgb)
```

## Recorte

```python
crop = image_rgb[500:1000, 1000:1500]
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(10, 8))
axes[0].imshow(image_rgb)
axes[1].imshow(crop)
```

## Translação

```python
# Dimensões da imagem original
height, width = image.shape[:2]

# Matriz de Translação
translation_x, translation_y = 800, 400
translation_matrix = np.float32([
    [1, 0, translation_x],
    [0, 1, translation_y]
])

#aplicando a matriz de transformação na imagem
translated = cv.warpAffine(image_rgb, translation_matrix,(width, height))

# Visualizando os resultados.
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(10, 8))
axes[0].imshow(image_rgb)
axes[1].imshow(translated)
```

## Rotação

```python
# Dimensões da imagem original
height, width = image.shape[:2]

# Definindo centro de rotação
center = (width // 2, height // 2)

# define o angulo de rotação
angle = 45 #rotação em 45 graus

# Define um operação de escala da imagem
scale = 1.0

# Deriva a matriz de rotação
rotation_matrix = cv.getRotationMatrix2D(center, angle, scale)

# Aplica a matriz de transformação na imagem.
rotated = cv.warpAffine(image_rgb, rotation_matrix, (width, height))

# Visualizar os resultados
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(10,8))
axes[0].imshow(image_rgb)
axes[1].imshow(rotated)
```

## Funções Utilitarias de Desenhos

```python
image_draw = cv.rectangle(image_rgb, (100,200), (500,500), (0,0,500), 5)
plt.imshow(image_draw)
```
