# Importações

```python
import numpy as np
import matplotlib.pylab as plt
import cv2 as cv
```

```python
image = cv.imread(r'images\planta-baixa.jpg')
image_rgb = cv.cvtColor(image, cv.COLOR_BGR2RGB)
plt.imshow(image_rgb)
```

# **Extração de caracteristicas**

## SIFT

```python
# Instanciando o detector SIFT
sift = cv.SIFT_create()

# Detectando keypoints e calculando os descritores
keypoints, descriptors = sift.detectAndCompute(image_rgb, None)

# Visualizando os keypoints
image_kp = cv.drawKeypoints(image_rgb, keypoints, None, flags=cv.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)

plt.figure(figsize=(4, 4), dpi=500)
plt.imshow(image_kp, interpolation='lanczos')
plt.axis('off')
plt.show()
```

## ORB

```python
# Instanciando o detector SIFT
sift = cv.ORB_create()

# Detectando keypoints e calculando os descritores
keypoints, descriptors = sift.detectAndCompute(image_rgb, None)

# Visualizando os keypoints
image_kp = cv.drawKeypoints(image_rgb, keypoints, None, flags=cv.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)

plt.figure(figsize=(4, 4), dpi=500)
plt.imshow(image_kp, interpolation='lanczos')
plt.axis('off')
plt.show()
```
